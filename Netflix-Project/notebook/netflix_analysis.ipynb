## Data Cleaning

```python
#Missing values check
df.isnull().sum()
```

```python
#Missing values fill

df['director'].fillna('Unknown', inplace=True)
df['cast'].fillna('Not Available', inplace=True)
df['country'].fillna('Unknown', inplace=True)
df['rating'].fillna('Not Rated', inplace=True)
```

```python
# Check karo missing values kam hue ya nahi
df.isnull().sum()
```

```python
# Missing Values Ab sirf:
# date_added    10
# duration       3

df['date_added'] = df['date_added'].fillna('Unknown')
df['duration'] = df['duration'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# Diplicae records check 
df.duplicated().sum()
```

```python
#duplicate remove
df.drop_duplicates(inplace=True)
```

# Visualization (Graphs)

```python
# Movies vs TV Shows
sns.countplot(x='type', data=df)
plt.title("Movies vs TV Shows")
plt.show()
```

### Observation

Netflix contains significantly more Movies than TV Shows.

```python
# Top Countries
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top Countries")
plt.show()
```

### Observation

United States contributes the highest amount of content followed by India and the United Kingdom.

```python
# Growth Over Years
df['release_year'].value_counts().sort_index().plot()
plt.title("Content Growth Over Years")
plt.show()
```

### Observation

Netflix content production increased rapidly after 2015 and peaked around 2019-2020.

## Insights & Conclusion

1. Netflix contains more Movies than TV Shows.

2. The United States is the largest content contributor on Netflix.

3. India is among the top content-producing countries.

4. Content production increased rapidly after 2015.

5. International TV Shows and Drama categories are highly popular.

6. Netflix expanded significantly in recent years by increasing global content.
